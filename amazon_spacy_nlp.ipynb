```python
# --- Amazon Reviews NER & Sentiment ---
import spacy
import pandas as pd
import matplotlib.pyplot as plt

# Load spaCy model
nlp = spacy.load('en_core_web_sm')

# Sample reviews dataset
data = pd.DataFrame({
    'review': [
        "This Samsung phone is amazing!",
        "Terrible experience with Apple charger.",
        "Sony headphones are good but a bit pricey.",
        "I love the new Dell laptop, very fast and reliable!",
        "This cheap adapter broke within a week, bad quality."
    ]
})

# Function for NER + Sentiment
def analyze_review(text):
    doc = nlp(text)
    entities = [(ent.text, ent.label_) for ent in doc.ents]
    positive_words = ["good", "amazing", "excellent", "love", "fast"]
    sentiment = "Positive" if any(word in text.lower() for word in positive_words) else "Negative"
    return entities, sentiment

data['entities'], data['sentiment'] = zip(*data['review'].apply(analyze_review))
print(data)

# Save output screenshot
fig, ax = plt.subplots(figsize=(8,2))
ax.axis('off')
ax.text(0, 0.5, str(data[['review','entities','sentiment']].head()), fontsize=10)
plt.savefig('../screenshots/spacy_ner_output.png')
plt.show()
# Save the processed data to a CSV file
```
